3. Data Cleaning: Outlier & String Methods
3.1 Deteksi Outlier Pakai Metode IQR
Selain cek missing value dan duplikat, cleaning yang lebih dalam juga perlu cek outlier, yaitu nilai yang jauh banget beda dari kebanyakan data, bisa jadi kesalahan input, bisa juga transaksi yang memang wajar tapi ekstrem.

Salah satu metode yang umum dipakai adalah IQR (Interquartile Range): hitung Q1 (persentil 25) dan Q3 (persentil 75), lalu nilai di luar rentang Q1 - 1.5×IQR sampai Q3 + 1.5×IQR dianggap outlier.

```python
finish_trans = df[df["status"]=="completed"].copy()

Q1 = finish_trans["total_sales"].quantile(0.25)
Q3 = finish_trans["total_sales"].quantile(0.75)
IQR= Q3-Q1
batas_bawah = Q1 - 1.5*IQR
batas_atas = Q3 + 1.5*IQR

print(f"Q1: {Q1:,.0f} | Q3: {Q3:,.0f} | IQR: {IQR:,.0f}")
print(f"Batas atas outlier: {batas_atas:,.0f}")

outlier = finish_trans[finish_trans["total_sales"] > batas_atas]
print("Jumlah transaksi outlier:", len(outlier))
print(f"Total pendapatan dari transaksi outlier: Rp {outlier['total_sales'].sum():,.0f}".replace(",", "."))
```

```python
finish_trans[finish_trans["total_sales"] < batas_bawah]
```

Penting: outlier itu nggak otomatis berarti "data salah". Di kasus kita ini kemungkinan besar itu transaksi pembelian dalam jumlah besar yang memang valid (misalnya buat acara atau reseller), bukan kesalahan input. Makanya, ketemu outlier itu bukan otomatis dihapus, tapi jadi bahan buat digali lebih lanjut.

4. GroupBy: Agregasi Ganda, Transform, dan Rank
4.1 .agg() dengan Banyak Fungsi Sekaligus
Kemarin kita cuma pakai satu fungsi agregasi tiap kali (.sum() atau .mean() sendiri-sendiri). Sekarang kita bisa hitung beberapa sekaligus pakai .agg().

```python
transaksi_selesai = df[df["status"] == "completed"].copy()

ringkasan_kategori = transaksi_selesai.groupby("category")["total_sales"].agg(["sum", "mean", "count"])
ringkasan_kategori.columns = ["total_pendapatan", "rata_rata_pendapatan", "jumlah_transaksi"]
ringkasan_kategori.round(0)
```

```python
ringkasan_kategori = transaksi_selesai.groupby("category")['total_sales'].agg(['sum','mean','count'])
ringkasan_kategori.columns = ['total_pendapatan','rata_rata_pendapatan','jumlah_transaksi']
ringkasan_kategori.reset_index()
```

4.2 .transform(): Bandingkan Tiap Baris dengan Rata-Rata Grupnya
groupby().sum() atau .mean() biasa itu hasilnya meringkas data (dari 9000 baris jadi cuma 3 baris kategori). Tapi kadang kita mau tetap punya 9000 baris itu, cuma nambahin kolom "rata-rata kategorinya berapa" di tiap baris. Nah, itu gunanya .transform().

```python
display(transaksi_selesai.groupby("category")["total_sales"].mean())

transaksi_selesai["rata2_kategori"] = transaksi_selesai.groupby("category")["total_sales"].transform("mean")
transaksi_selesai
```

```python
transaksi_selesai["rata2_kategori"] = (
    transaksi_selesai.groupby("category")["total_sales"].transform("mean")
)
transaksi_selesai["selisih_dari_rata2"] = (
    transaksi_selesai["total_sales"] - transaksi_selesai["rata2_kategori"]
)

transaksi_selesai[["product_name", "category", "total_sales", "rata2_kategori", "selisih_dari_rata2"]].head()
```

Kolom selisih_dari_rata2 ini berguna banget buat cari transaksi yang jauh di atas atau di bawah kebiasaan kategorinya, kita pakai ini di case study nanti.

4.3 .rank(): Bikin Peringkat di Dalam Grup
.rank() dipakai buat kasih peringkat, dan kalau digabung sama groupby(), kita bisa bikin peringkat di dalam masing-masing grup, bukan peringkat keseluruhan.

```python
transaksi_selesai.groupby(["category", "product_name"])["total_sales"].sum().reset_index()
```

```python
pendapatan_produk = (transaksi_selesai.groupby(["category","product_name"])["total_sales"].sum().reset_index())

pendapatan_produk["rank_dalam_kategori"]=(pendapatan_produk.groupby("category")["total_sales"].rank(ascending=False, method="min"))

pendapatan_produk.sort_values(["category","rank_dalam_kategori"]).head(10)
```

Pertanyaan 2: Seberapa besar kontribusi transaksi "outlier" (pembelian besar) terhadap total pendapatan toko?

```python
# Hitung kontribusi pendapatan transaksi outlier
total_pendapatan_toko = transaksi_selesai["total_sales"].sum()
total_pendapatan_outlier = outlier["total_sales"].sum()
kontribusi_outlier = total_pendapatan_outlier / total_pendapatan_toko * 100

print(f"Total pendapatan toko: Rp{total_pendapatan_toko:,.0f}".replace(",", "."))
print(f"Pendapatan dari outlier: Rp{total_pendapatan_outlier:,.0f}".replace(",", "."))
print(f"Kontribusi outlier: {kontribusi_outlier:.2f}%")

```
